```python
from flask import Flask, render_template

app = Flask(__name__)


@app.route("/")
def hello_world():
    return render_template("index.html")
```

The route above returns `render_template("index.html")` rather than a raw HTML string. Keeping the markup in `templates/index.html` avoids managing long HTML strings inside Python code.

```python
# Determine whether this notebook is running in Jupyter or Google Colab
def get_shell_name():
    try:
        return get_ipython().__class__.__name__
    except NameError:
        return "standard Python interpreter"


shell = get_shell_name()
if shell == "ZMQInteractiveShell":
    print("Running in Jupyter")
elif shell == "Shell":
    print("Running in Google Colab")
else:
    print(f"Running in: {shell}")
```

```python
# Start the development server (stop the cell to shut it down)
app.run(host="127.0.0.1", port=5000)
```
